# 01 · Hình schematic (đợt 0) — không cần dữ liệu, không GPU

Ba hình: **F3** pipeline (mốc thời gian đọc từ `schedule.py`),
**F2** sơ đồ khái niệm, **A1** bảng phân quyền nhãn.

Cả ba là hình **vẽ**, nên mang nhãn `schematic` — không được gắn `R0`/`R1+R2`,
vì chúng không chứa số đo nào.

```python
# Chạy từ thư mục gốc repo. Notebook CHỈ gọi tools/figures.py —
# không tự viết lại logic vẽ, để hình trong notebook và hình trong báo cáo
# luôn là cùng một hình.
import json, os, subprocess, sys
from pathlib import Path

ROOT = Path.cwd()
assert (ROOT / 'tools' / 'figures.py').exists(), 'hãy chạy notebook từ thư mục gốc repo'
FIG = Path(os.environ.get('DIRE_FIG_OUT', 'figs'))

def fig_cmd(*args):
    """Chạy tools/figures.py, in lệnh rồi in đầu ra — lệnh in ra là lệnh tái lập."""
    cmd = [sys.executable, str(ROOT / 'tools' / 'figures.py'), *map(str, args)]
    print('$', ' '.join(cmd[1:]))
    r = subprocess.run(cmd, cwd=ROOT, capture_output=True, text=True)
    print(r.stdout.strip() or r.stderr.strip())
    r.check_returncode()
    return r

def sidecar(name):
    """Đọc sidecar: nguồn, n, nhãn. Không có sidecar = hình không kiểm được."""
    return json.loads((FIG / 'data' / f'{name}.json').read_text(encoding='utf-8'))
```

## F3 — pipeline

`--steps` phải khớp giao thức: bài báo dùng **S=20** cho DIRE.

```python
fig_cmd('f3-pipeline', '--out', FIG, '--steps', 20)
sidecar('f3-pipeline')
```

Ô kiểm: mốc cuối phải là 950, không phải 999 — sai chỗ này là sai 25× chi phí.

```python
sc = sidecar('f3-pipeline')
assert sc['timesteps'][0] == 0 and sc['timesteps'][-1] == 950, sc['timesteps']
assert sc['nfe_per_image'] == 40
print('mốc:', sc['timesteps'])
```
